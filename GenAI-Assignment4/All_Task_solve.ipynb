# Task 1 — Write Sales Records to a File

```python
sales = [1200, 450, 980, 1500, 3000]

with open("sales_data.txt", "w") as file:
    for sale in sales:
        file.write(str(sale) + "\n")

with open("sales_data.txt", "r") as file:
    print(file.read())
```

# Task 2 — Read File in Different Ways

```python
with open("sales_data.txt", "r") as file:
    print("Using read():")
    print(file.read())

# Read the first line
with open("sales_data.txt", "r") as file:
    print("Using readline():")
    print(file.readline().strip())

# Read all lines and convert them into integers
with open("sales_data.txt", "r") as file:
    lines = file.readlines()

sales_list = []

for line in lines:
    sales_list.append(int(line.strip()))

print("Integer list:", sales_list)
```

# Task 3 — Append New Sales

```python
new_sales = [5000, 2500, 1700]

with open("sales_data.txt", "a") as file:
    for sale in new_sales:
        file.write(str(sale) + "\n")

with open("sales_data.txt", "r") as file:
    print(file.read())

# Extra: Count total lines
with open("sales_data.txt", "r") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
```

# Task 4 — Generate Summary Report from File

```python
sales = []

with open("sales_data.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    sales.append(int(line.strip()))

total_sales = sum(sales)
highest_sale = max(sales)
lowest_sale = min(sales)
average_sale = total_sales / len(sales)

print("Total Sales:", total_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
print("Average Sale:", average_sale)
```

# Task 5 — Create Product Info File

```python
with open("products.txt", "w") as file:
    for i in range(3):
        name = input("Enter product name: ")
        price = input("Enter product price: ")

        file.write(name + " | " + price + "\n")

with open("products.txt", "r") as file:
    print("\nProduct Information:")

    for line in file:
        print(line.strip())
```

# Task 6 — Read File Safely

```python
import os

filename = input("Enter filename: ")

if os.path.exists(filename):
    with open(filename, "r") as file:
        print(file.read())
else:
    print("File not found. Please check the filename.")
```

# Task 7 — Mini Project: Export Discounted Prices

```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount_percent = float(input("Enter discount percentage: "))

total_discounted_price = 0
total_items = 0

with open("discount_report.txt", "w") as file:
    file.write("Product | Original Price | Discounted Price\n")

    for product, price in prices.items():
        discount = price * discount_percent / 100
        discounted_price = price - discount

        file.write(
            product + " | " + str(price) + " | "
            + str(discounted_price) + "\n"
        )

        total_discounted_price += discounted_price
        total_items += 1

    average_price = total_discounted_price / total_items

    file.write("\nTotal Items: " + str(total_items) + "\n")
    file.write(
        "Average Discounted Price: " + str(average_price) + "\n"
    )

with open("discount_report.txt", "r") as file:
    print(file.read())
```
